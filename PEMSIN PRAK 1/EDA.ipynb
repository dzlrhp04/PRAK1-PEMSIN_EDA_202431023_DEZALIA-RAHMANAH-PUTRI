## Adjustmen of Variabel Types

```python
df.dtypes #buat ngelihat type data dari setiap kolom

##202431023_Dezalia Rahmanah Putri
```

```python
cat_col = ["gender", "ever_married", "work_type", "Residence_type", "smoking_status"]

for col in cat_col:
    df[col] = df[col].astype('category')

df.dtypes

##202431023_Dezalia Rahmanah Putri
```

```python
df.isnull().sum() #untuk menjumlahkan semua nilai dari kolom yang ada, lalu ini buat nyari ada ga di kolomnya data yg kosong

##202431023_Dezalia Rahmanah Putri
```

```python
df['bmi'] = df['bmi'].fillna(df['bmi'].median())

df.head()

##202431023_Dezalia Rahmanah Putri
```

```python
df.isnull().sum()

##202431023_Dezalia Rahmanah Putri
```

```python
sns.boxplot(x="bmi", data = df)
plt.title("Distribusi Kelas BMI")
plt.show()

##202431023_Dezalia Rahmanah Putri
```

```python
Q1 = df['age'].quantile(1/4) #mengambil 25% data (batas bawah)
Q3 = df['age'].quantile(3/4) #mengambil 75% data (batas atas)

#untuk menentukan batas bawah atasnya kita memerlukn nilai quantile
IQR = Q3 - Q1

lower_bound = Q1 - 3/2 * IQR
upper_bound = Q3 + 3/2 * IQR

outliers = df[(df["age"] < lower_bound) | (df["age"] > upper_bound)]
print ("jumlah outlier umur: ", len(outliers))

##202431023_Dezalia Rahmanah Putri
```
